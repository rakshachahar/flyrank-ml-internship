## 2. Intended use and limits

*Who uses this, for what — and where it stops being valid.*

This playbook is designed for content editors and SEO teams to support content-refresh decisions. It provides ranked recommendations based on historical search-performance signals.

The recommendations should not be interpreted as guarantees of improved rankings or automated publishing decisions. Human judgment remains essential.

```python
use_case = {
    "Primary Users": "Content Editors and SEO Teams",
    "Purpose": "Decision Support",
    "Automation": "Not Fully Automated"
}

print(use_case)
```

## 3. Human review + the no-go list

*What a person must check before acting. What should never be automated.*

Before acting on any recommendation, reviewers should verify page quality, business importance, recent content updates, and any external factors that may affect performance.

The following should never be automated:
- Publishing content changes
- Deleting website pages
- Making business-critical SEO decisions
- Using client-identifying information

```python
review_items = [
    "Check page quality",
    "Verify business importance",
    "Review recent updates",
    "Confirm recommendation manually"
]

for item in review_items:
    print("✓", item)
```

## 4. Monitoring / retrain triggers

*What would tell you the recommendations went stale?*

The recommendation system should be reviewed whenever search-performance patterns change significantly, new historical data becomes available, or overall model performance declines.

Regular evaluation helps ensure that recommendations remain useful and aligned with current search-performance observations.

```python
triggers = [
    "Performance decline",
    "New historical data",
    "Search pattern changes",
    "Periodic model evaluation"
]

print(triggers)
```
