## Title
Machine Learning-Based Prioritization of Website Content Refresh Using Search Performance Signals

### 1. Problem (1 min)
Explain why reviewing every website page manually is inefficient and how content teams need a data-driven way to prioritize pages for refresh.

### 2. Method (1 min)
Introduce the FlyRank anonymized search-performance dataset, describe the selected features (impressions, clicks, CTR, average position), explain the baseline rule, and the Random Forest classifier.

### 3. Results (1 min)
Show the comparison chart between the baseline and the Random Forest model. Explain that the machine learning model produced stronger prioritization while using the same validation strategy.

### 4. Recommendation (1 min)
Discuss the action playbook:
- Refresh declining pages.
- Improve CTR through title and meta description optimization.
- Continue monitoring stable pages.

### 5. Key Takeaway (1 min)
The model supports human decision-making by prioritizing pages for review using observable search-performance signals. It is intended as decision-support rather than automated publishing.
